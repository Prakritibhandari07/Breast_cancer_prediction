```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn.datasets
from sklearn.model_selection import train_test_split
```

```python
#loading the data from sklearn
breast_cancer_dataset=sklearn.datasets.load_breast_cancer()
```

```python
data_frame=pd.DataFrame(breast_cancer_dataset.data,columns=breast_cancer_dataset.feature_names)
```

```python
data_frame['label']=breast_cancer_dataset.target
```

separating the features and target

```python
X=data_frame.drop(columns='label')
#X = data_frame.drop('label', axis=1)
Y=data_frame['label']
```

```python
print(X)
```

```python
print(Y)
```

Splitting the data into training data and test data

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=2)
```

```python
print(X.shape,X_train.shape,X_test.shape)
```

```python
print(Y.shape,Y_train.shape,Y_test.shape)
```

Standarize the data

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()

X_train_std=scaler.fit_transform(X_train)

X_test_std=scaler.transform(X_test)
```

```python
print(X_train_std)
```

Building the neural network

```python
pip install tensorflow
```

```python
#importing tensorflow and keras
import tensorflow as tf
tf.random.set_seed(3)
from tensorflow import keras
```

```python
#setting up the layers of Neural Network
model=keras.Sequential([keras.Input(shape=(30,)),
                        keras.layers.Flatten(),
                        keras.layers.Dense(20, activation='relu'),
                        keras.layers.Dense(20, activation='relu'),
                        keras.layers.Dense(2, activation='softmax')
                       ])
```

```python
#compiling the heural network

model.compile(optimizer='adam',
             loss='sparse_categorical_crossentropy',
             metrics=['accuracy'])
```

```python
#training the neural network

history=model.fit(X_train_std,Y_train,validation_split=0.1,epochs=10)
```

visualizing accuracy and loss

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])

plt.title('model_accuracy')
plt.ylabel('accuracy')
plt.xlabel('epoch')

plt.legend(['training_data','validation_data'],loc='lower right')
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])

plt.title('model_loss')
plt.ylabel('loss')
plt.xlabel('epoch')

plt.legend(['training_data','validation_data'],loc='lower right')
```

Accuracy of the model on test data

```python
loss,accuracy=model.evaluate(X_test_std,Y_test)
print(accuracy)
```

Building the predictive system

```python
input_data=(11.52,18.75,73.34,409,0.09524,0.05473,0.03036,0.02278,0.192,0.05907,0.3249,0.9591,2.183,23.47,0.008328,0.008722,0.01349,0.00867,0.03218,0.002386,12.84,22.47,81.81,506.2,0.1249,0.0872,0.09076,0.06316,0.3306,0.07036
)

#change the input data to numpy array

input_data_as_numpy_array=np.asarray(input_data)

#reshape the numpy array as we are predicting for one data point
input_data_reshaped=input_data_as_numpy_array.reshape(1,-1)

#standardizing the input data

input_data_std=scaler.transform(input_data_reshaped)

prediction=model.predict(input_data_std)
print(prediction)

prediction_label=[np.argmax(prediction)]
print(prediction_label)

if(prediction_label[0]==0):
    print("The tumor is Malignant")
else:
    print("The tumor is Benign")
```
